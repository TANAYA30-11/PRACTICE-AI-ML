```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

#Matplotlib Basic


```python
x=np.linspace(0,5,11)
x
y=x**3
y
```

```python
plt.plot(x,y)
plt.xlabel('X Label')
plt.ylabel('Y Label')
plt.title('hello')
```

#Subplots in Matplotlib


```python
plt.subplot(2,2,1)
plt.plot(y,x)
plt.subplot(2,2,2)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(x,y)
plt.subplot(2,2,4)
plt.plot(x,y)

```

#Object oriented way

```python
fig=plt.figure()
#fig.add_axes([0.1,0.1,1,1])
axis1=fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)
axis2=fig.add_axes([.2,.7,.3,.3])
axis2.plot(x,y)
axis3=fig.add_axes([.6,.7,.3,.3])
axis3.plot(x,y)

```

#types of plots

```python
plt.scatter(x,y)
```

```python
plt.hist(x)
```

```python
plt.boxplot(x)
```

#Saving plots

```python
fig=plt.figure()
axis=fig.add_axes([0.1,0.1,1,1])
axis.plot(x,y)
plt.savefig('my_plot.png')
```

#Working with images

```python
import matplotlib.image as mpimg
```

```python
img=mpimg.imread('my_plot.png')
plt.imshow(img)
plt.axis('off')
```

```python
cropped_img=img[100:200,300:400]
plt.imshow(cropped_img)
plt.axis('off')
```
